# Generalized Regression Neural Network (GRNN)
GRNN gives nearby training points larger weights. The final answer is a weighted average.

```python
import math

def grnn_prediction(training_x, training_y, test_x, sigma):
    weights = []
    for training_point in training_x:
        squared_distance = 0
        for i in range(len(test_x)):
            difference = test_x[i] - training_point[i]
            squared_distance = squared_distance + difference ** 2
        weight = math.exp(-squared_distance / (2 * sigma ** 2))
        weights.append(weight)

    numerator = 0
    denominator = 0
    for i in range(len(weights)):
        numerator = numerator + weights[i] * training_y[i]
        denominator = denominator + weights[i]

    prediction = numerator / denominator
    return prediction, weights
```

## One-dimensional example

```python
x_train = [(2,), (4,), (6,), (8,)]
y_train = [3, 5, 7, 9]
test_point = (5,)
sigma = 1
answer, weights = grnn_prediction(x_train, y_train, test_point, sigma)

rounded_weights = []
for value in weights:
    rounded_weights.append(round(value, 4))
print('Weights:', rounded_weights)
print('Predicted output:', round(answer, 4))
```

## Two-dimensional example

```python
x_train_2d = [(1, 2), (2, 3), (3, 4)]
y_train_2d = [3, 6, 9]
test_point_2d = (2, 2)
answer_2d, weights_2d = grnn_prediction(x_train_2d, y_train_2d, test_point_2d, 1)

rounded_weights_2d = []
for value in weights_2d:
    rounded_weights_2d.append(round(value, 4))
print('Weights:', rounded_weights_2d)
print('Predicted output:', round(answer_2d, 4))
```
