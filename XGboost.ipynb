# XGBoost
This small example builds one classification split and updates the predicted probabilities.

```python
import math

x = [2, 3, 7, 8]
y = [1, 1, 0, 0]
predicted_scores = [0, 0, 0, 0]
regularization = 1
gamma = 0
split_value = 5

def sigmoid(score):
    answer = 1 / (1 + math.exp(-score))
    return answer
```

## Step 1: Calculate probabilities, gradients, and Hessians

```python
probabilities = []
gradients = []
hessians = []
for i in range(len(predicted_scores)):
    probability = sigmoid(predicted_scores[i])
    gradient = probability - y[i]
    hessian = probability * (1 - probability)
    probabilities.append(probability)
    gradients.append(gradient)
    hessians.append(hessian)

print('Probabilities:', probabilities)
print('Gradients:', gradients)
print('Hessians:', hessians)
```

## Step 2: Split at x < 5 and calculate the leaf weights

```python
left_indices = []
right_indices = []
for i in range(len(x)):
    if x[i] < split_value:
        left_indices.append(i)
    else:
        right_indices.append(i)

left_gradient = 0
left_hessian = 0
for i in left_indices:
    left_gradient = left_gradient + gradients[i]
    left_hessian = left_hessian + hessians[i]

right_gradient = 0
right_hessian = 0
for i in right_indices:
    right_gradient = right_gradient + gradients[i]
    right_hessian = right_hessian + hessians[i]

left_weight = -left_gradient / (left_hessian + regularization)
right_weight = -right_gradient / (right_hessian + regularization)
print('Left leaf weight:', round(left_weight, 3))
print('Right leaf weight:', round(right_weight, 3))
```

## Step 3: Calculate the split gain

```python
total_gradient = left_gradient + right_gradient
total_hessian = left_hessian + right_hessian
left_gain = left_gradient ** 2 / (left_hessian + regularization)
right_gain = right_gradient ** 2 / (right_hessian + regularization)
old_gain = total_gradient ** 2 / (total_hessian + regularization)
gain = 0.5 * (left_gain + right_gain - old_gain) - gamma
print('Split gain:', round(gain, 3))
```

## Step 4: Update the predictions

```python
new_scores = []
new_probabilities = []
for value in x:
    if value < split_value:
        new_score = left_weight
    else:
        new_score = right_weight
    new_scores.append(new_score)
    new_probabilities.append(sigmoid(new_score))

rounded_scores = []
rounded_probabilities = []
for value in new_scores:
    rounded_scores.append(round(value, 3))
for value in new_probabilities:
    rounded_probabilities.append(round(value, 3))
print('Updated scores:', rounded_scores)
print('Updated probabilities:', rounded_probabilities)
```
