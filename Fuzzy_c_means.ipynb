# Fuzzy C-means
This algorithm lets every data point belong partly to each cluster.

```python
data = [1, 2, 7, 8]
number_of_clusters = 2
m = 2
maximum_iterations = 100
tolerance = 0.00001

membership = [[0.6, 0.4], [0.7, 0.3], [0.3, 0.7], [0.2, 0.8]]
```

## Repeat the centroid and membership calculations

```python
for iteration in range(maximum_iterations):
    centroids = []
    for cluster in range(number_of_clusters):
        numerator = 0
        denominator = 0
        for i in range(len(data)):
            fuzzy_weight = membership[i][cluster] ** m
            numerator = numerator + fuzzy_weight * data[i]
            denominator = denominator + fuzzy_weight
        centroids.append(numerator / denominator)

    new_membership = []
    for point in data:
        distances = []
        for centroid in centroids:
            distances.append(abs(point - centroid))

        row = []
        if 0 in distances:
            exact_cluster = distances.index(0)
            for cluster in range(number_of_clusters):
                if cluster == exact_cluster:
                    row.append(1.0)
                else:
                    row.append(0.0)
        else:
            for cluster in range(number_of_clusters):
                total = 0
                for other_cluster in range(number_of_clusters):
                    ratio = distances[cluster] / distances[other_cluster]
                    total = total + ratio ** (2 / (m - 1))
                row.append(1 / total)
        new_membership.append(row)

    largest_change = 0
    for i in range(len(data)):
        for cluster in range(number_of_clusters):
            change = abs(new_membership[i][cluster] - membership[i][cluster])
            if change > largest_change:
                largest_change = change

    membership = new_membership
    if largest_change < tolerance:
        break

print('Iterations:', iteration + 1)
print('Centroids:')
for centroid in centroids:
    print(round(centroid, 4))

print('Membership matrix:')
for i in range(len(data)):
    rounded_row = []
    for value in membership[i]:
        rounded_row.append(round(value, 4))
    print(data[i], rounded_row)
```
