# Modified K-means
Instead of selecting the first centroids by chance, modified k-means uses the two points that are farthest apart.

```python
import math

points = [(3, 4), (6, 7), (4, 5), (5, 7), (2, 6)]
k = 2

def distance(point1, point2):
    x_difference = point1[0] - point2[0]
    y_difference = point1[1] - point2[1]
    answer = math.sqrt(x_difference ** 2 + y_difference ** 2)
    return answer
```

## Step 1: Find the two farthest points

```python
largest_distance = -1
centroid1 = None
centroid2 = None

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        current_distance = distance(points[i], points[j])
        if current_distance > largest_distance:
            largest_distance = current_distance
            centroid1 = points[i]
            centroid2 = points[j]

centroids = [centroid1, centroid2]
print('Starting centroids:', centroids)
print('Largest distance:', round(largest_distance, 2))
```

## Step 2: Assign points and update the centroids

```python
for iteration in range(10):
    cluster1 = []
    cluster2 = []

    for point in points:
        distance1 = distance(point, centroids[0])
        distance2 = distance(point, centroids[1])
        if distance1 <= distance2:
            cluster1.append(point)
        else:
            cluster2.append(point)

    cluster1_x_total = 0
    cluster1_y_total = 0
    for point in cluster1:
        cluster1_x_total = cluster1_x_total + point[0]
        cluster1_y_total = cluster1_y_total + point[1]

    cluster2_x_total = 0
    cluster2_y_total = 0
    for point in cluster2:
        cluster2_x_total = cluster2_x_total + point[0]
        cluster2_y_total = cluster2_y_total + point[1]

    new_centroid1 = (cluster1_x_total / len(cluster1), cluster1_y_total / len(cluster1))
    new_centroid2 = (cluster2_x_total / len(cluster2), cluster2_y_total / len(cluster2))
    new_centroids = [new_centroid1, new_centroid2]
    print('Iteration', iteration + 1, new_centroids)

    if new_centroids == centroids:
        break
    centroids = new_centroids

print('Final cluster 1:', cluster1)
print('Final cluster 2:', cluster2)
```

## Step 3: Classify a new point

```python
new_point = (2, 7)
distance1 = distance(new_point, centroids[0])
distance2 = distance(new_point, centroids[1])
if distance1 <= distance2:
    print(new_point, 'belongs to Cluster 1')
else:
    print(new_point, 'belongs to Cluster 2')
```
